## 1. The "Sequel Bias" & Episode Length vs. Score

**Question:** Do long-running shows or sequels systematically score higher because only invested fans rate them?  


```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Define episode bins (e.g., 1 cour, 2 cour, 4 cour, and long-running)
bins = [0, 13, 26, 50, np.inf]
labels = ['1-13 eps (1 Cour)', '14-26 eps (2 Cour)', '27-50 eps', '50+ eps']

# Create a new categorical column for episode bins
df['episode_bin'] = pd.cut(df['Episodes'], bins=bins, labels=labels)

# Plot
plt.figure(figsize=(10, 6))
sns.boxplot(x='episode_bin', y='Score', hue='episode_bin', data=df, palette='Set2', legend=False)

plt.title('The Survivor Bias: Episode Count vs. Score', fontsize=14)
plt.xlabel('Episode Length')
plt.ylabel('Score')
sns.despine()
plt.show()
```

**Question:** How has the share of classic genres (like Mecha, Fantasy, Romance, Sci-Fi) shifted over the decades?  


```python
# 1. Create a decade column (using the air_year you parsed earlier)
df['Decade'] = (df['air_year'] // 10) * 10

# 2. Explode the Genres column so each anime has 1 row per genre
# (Assuming genres are separated by commas)
df_genres = df.assign(Genres=df['Genres'].astype(str).str.split(', ')).explode('Genres')

# 3. Filter for a few key genres to make the chart readable
target_genres = ['Mecha', 'Fantasy', 'Romance', 'Sci-Fi', 'Slice of Life', 'Action']
df_filtered = df_genres[df_genres['Genres'].isin(target_genres)]

# 4. Filter for 1990s to 2020s and reset the index to fix the duplicate label error
df_filtered = df_filtered[df_filtered['Decade'] >= 1990].reset_index(drop=True)

# 5. Create a crosstab and normalize it to get percentages
genre_trends = pd.crosstab(df_filtered['Decade'], df_filtered['Genres'], normalize='index') * 100

# Plot
genre_trends.plot(kind='bar', stacked=True, figsize=(12, 6), colormap='viridis', width=0.85)

```

Throughout the decades, Fantasy has seen a sharp increase in popularity due to the boom of Isekai anime. 

## 3. Source Material Premium

**Question:** Do original anime score worse on average because they lack an established fan base, or do adaptations perform better?  
**Method:** We use a violin plot to see the density and distribution of scores grouped by the top source materials.

```python
# Get the top 6 most common source materials for a cleaner chart
top_sources = df['Source'].value_counts().nlargest(6).index
df_source = df[df['Source'].isin(top_sources)]

# Plot
plt.figure(figsize=(12, 6))
sns.violinplot(x='Source', y='Score', data=df_source, hue='Source', palette='muted', order=top_sources, inner='quartile', legend=False)
plt.title('Source Material Premium: Score Distribution by Source', fontsize=14)
plt.xlabel('Source Material')
plt.ylabel('Score')
sns.despine()
plt.show()

# Create the Summary Table
source_stats = df_source.groupby('Source').agg(
    Mean_Score=('Score', 'mean'),
    Median_Score=('Score', 'median'),
    Variance=('Score', 'var'),
    Count=('Score', 'count')
).sort_values(by='Count', ascending=False).round(2)

display(source_stats)
```

## Summary Findings
- **Sequel/Survivor Bias:** Shows with 50+ episodes generally have a higher median score. This is likely because only successful, highly-rated shows survive long enough to get that many episodes, and only dedicated fans continue rating them deep into the run.

- **Genre Shifts:** (Analyze your bar chart output here—e.g., *Mecha peaked in the 90s while Fantasy/Isekai has eaten up market share in the 2010s/2020s.*)

- **Source Premium:** Adaptations from Manga and Light Novels tend to have slightly higher median scores than Original anime, but Original anime exhibit much higher variance (they are more "hit or miss"). Adaptations from Visual Novels have a lower mean score from all sources other than unknown, while also having the lowest variance. This stems from the fact that Visual Novels often have different choices, endings, and deeper insights into characters and stories that are difficult to translate to an anime format. 


## Creating Models


```python
import pandas as pd

def get_best_anime_matches(anime_name, user_matrix, metadata_df, min_overlap=20, top_n=5):
    """
    Recommends anime based on highly correlated user ratings (similar fans) 
    AND ensures the recommendations share at least one genre with the target anime.
    """
    
    # 1. Check if the anime exists in our matrix
    if anime_name not in user_matrix.columns:
        return f"Sorry, '{anime_name}' was not found in the ratings database."
    
    # 2. Get the target anime's ratings and its genres
    target_ratings = user_matrix[anime_name]
    
    # Extract genres for the target anime and split them into a list
    target_genres_str = metadata_df.loc[metadata_df['Name'] == anime_name, 'Genres'].values[0]
    target_genres = [g.strip() for g in str(target_genres_str).split(',')]
    
    # 3. Calculate correlation with all other anime (The "Similar Fans" part)
    similar_anime = user_matrix.corrwith(target_ratings, min_periods=min_overlap)
    corr_df = pd.DataFrame(similar_anime, columns=['Correlation']).dropna()
    
    # Remove the target anime itself from the results
    corr_df = corr_df.drop(index=anime_name, errors='ignore')
    
    # 4. Join the metadata (Genres, Members, Score, etc.)
    # We temporarily set the index to 'Name' to make joining easy
    metadata_indexed = metadata_df.set_index('Name')
    final_df = corr_df.join(metadata_indexed[['Genres', 'Score', 'Members']])
    
    # 5. Filter for Genre Overlap (The "Genre Match" part)
    # We check if ANY of the genres in the recommended anime match the target's genres
    def shares_genre(genre_string):
        if pd.isna(genre_string): return False
        anime_genres = [g.strip() for g in str(genre_string).split(',')]
        # Returns True if there is an intersection between the two lists
        return len(set(target_genres).intersection(set(anime_genres))) > 0
    
    final_df['Shares_Genre'] = final_df['Genres'].apply(shares_genre)
    final_df = final_df[final_df['Shares_Genre'] == True]
    
    # 6. Sort by highest correlation and return the top N
    best_matches = final_df.sort_values(by='Correlation', ascending=False)
    
    print(f"Top {top_n} Recommendations for fans of '{anime_name}':")
    print(f"Target Genres: {', '.join(target_genres)}\n")
    
    return best_matches[['Correlation', 'Genres', 'Score', 'Members']].head(top_n)
```
